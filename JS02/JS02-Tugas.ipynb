**Deskripsi Tugas**

Pada tugas pratikum ini Anda akan menggunakan data "Wisconsin Breast Cancer". Data tersebut terdiri dari 569 data yang digunakan untuk mendiagnonis jenis kanker Malignant (M) dan Benign (B). Tugas Anda adalah,

    Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

    Lakukan proses encoding pada kolom "diagnosis".

    Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

Import dataset

```python
df = pd.read_csv('wbc.csv')
```

1. Pisahkan variabel yang dapat digunakan dan yang tidak dapat digunakan

```python
# Kolom 'id' tidak berguna untuk prediksi medis.
# Seringkali ada kolom 'Unnamed: 32' yang kosong karena koma berlebih di dataset.
cols_to_drop = ['id']
if 'Unnamed: 32' in df.columns:
    cols_to_drop.append('Unnamed: 32')

df = df.drop(columns=cols_to_drop)
print("Langkah 1 Selesai: Kolom yang tidak berguna telah dihapus.")
```

ketika di run lagi data sudah tidak dapat ditemukan maka akan terjadi error

2. Lakukan proses encoding pada kolom "diagnosis"

```python
# Mengubah 'M' (Malignant) dan 'B' (Benign) menjadi angka 1 dan 0
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
print("\nLangkah 2 Selesai: Kolom 'diagnosis' telah di-encode.")
print("Mapping Label:", dict(zip(le.classes_, le.transform(le.classes_))))
```

3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik

```python
#  memisahkan fitur numerik (X) dengan target (y)
# Biasanya target ('diagnosis') TIDAK ikut di-standardisasi karena dia adalah output prediksi (0 atau 1)
# Mengambil semua nama kolom numerik KECUALI 'diagnosis'
fitur_numerik = df.columns.drop('diagnosis')
# Inisialisasi StandardScaler
scaler = StandardScaler()
# Melakukan standardisasi hanya pada fitur numeriknya saja
df[fitur_numerik] = scaler.fit_transform(df[fitur_numerik])
print("\nLangkah 3 Selesai: Kolom numerik telah distandardisasi.")
```
